```python
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
import io

filename = list(uploaded.keys())[0]

df = pd.read_csv(io.BytesIO(uploaded[filename]))

print("Dataset loaded successfully!")
print("Rows and Columns:", df.shape)

df.head()
```

```python
df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace("Â£", "", regex=False)
    .str.replace("£", "", regex=False)
    .str.strip()
)

df["Price"] = pd.to_numeric(df["Price"], errors="coerce")

print(df["Price"].head())
```

```python
average_price = df["Price"].mean()
highest_price = df["Price"].max()
lowest_price = df["Price"].min()

print("Average Price:", round(average_price, 2))
print("Highest Price:", round(highest_price, 2))
print("Lowest Price:", round(lowest_price, 2))
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df["Price"].dropna(), bins=10)

plt.title("Distribution of Book Prices")
plt.xlabel("Price")
plt.ylabel("Number of Books")

plt.show()
```

```python
top_books = df.nlargest(10, "Price")

plt.figure(figsize=(10, 6))

plt.barh(top_books["Book Name"], top_books["Price"])

plt.title("Top 10 Most Expensive Books")
plt.xlabel("Price")
plt.ylabel("Book Name")

plt.gca().invert_yaxis()

plt.show()
```

```python
df.to_csv("cleaned_books_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```
